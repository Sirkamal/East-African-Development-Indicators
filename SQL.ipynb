```python
import pandas as pd
import sqlite3

# Load your clean CSV
df = pd.read_csv(r'C:\Users\abdin\Desktop\Decision Analysist\Chapter 1 Projects\East African Countries Development Indicators\Data\EAC_Development_Indicators_Clean.csv')

# Create an in-memory database
conn = sqlite3.connect(':memory:')

# Load the dataframe into the database as a table
df.to_sql('eac_indicators', conn, index=False, if_exists='replace')

print("Database ready. Rows loaded:", len(df))
```

```python
query7 = """
SELECT "Country Name",
       COUNT("Value") as years_with_data,
       AVG("Value") as average_gdp,
       MAX("Value") as highest_gdp,
       MIN("Value") as lowest_gdp
FROM eac_indicators
WHERE "Series Name" = 'GDP per capita (current LCU)'
GROUP BY "Country Name"
ORDER BY average_gdp DESC
"""

result7 = pd.read_sql_query(query7, conn)
result7
```

```python
query8 = """
SELECT "Country Name", 
       COUNT("Value") as years_with_data
FROM eac_indicators
WHERE "Series Name" = 'Unemployment, total (% of total labor force) (national estimate)'
GROUP BY "Country Name"
HAVING years_with_data < 10
ORDER BY years_with_data DESC
"""

result8 = pd.read_sql_query(query8, conn)
result8
```

```python
# Create two seperate tables from your data

gdp_table = df[df['Series Name'] == 'GDP per capita (current LCU)'][
    ["Country Name", "Year", "Value"]
].rename(columns={"Value": "GDP per capita"})

pop_table = df[df['Series Name'] == 'Population, total'][
    ["Country Name", "Year", "Value"]
].rename(columns={"Value": "Population"})

gdp_table.to_sql('gdp_table', conn, index=False, if_exists='replace')
pop_table.to_sql('pop_table', conn, index=False, if_exists='replace')   

print("GDP rows:", len(gdp_table))
print("Population rows:", len(pop_table))
```

```python
query9 = """
SELECT g."Country Name", 
         g."Year", 
         g."GDP per capita", 
         p."Population"
FROM gdp_table g
JOIN pop_table p
    ON g."Country Name" = p."Country Name" 
    AND g."Year" = p."Year"
ORDER BY g."Country Name", g."Year"
"""

result9 = pd.read_sql_query(query9, conn)
result9
```

```python
query10 = """
SELECT 
    "Country Name",
    AVG(Value) AS avg_gdp_per_capita_2020_2024
FROM eac_indicators
WHERE 
    "Series Name" = 'GDP per capita (current LCU)'
    AND Year BETWEEN 2020 AND 2024
    AND Value IS NOT NULL
GROUP BY "Country Name"
ORDER BY avg_gdp_per_capita_2020_2024 DESC
"""

result10 = pd.read_sql_query(query10, conn) 
result10
```

```python
query11 = """
SELECT 
    "Country Name",
    "Year",
    "Value" as expenditure_pct_gdp,
    RANK() OVER (
        PARTITION BY "Year" 
        ORDER BY "Value" DESC
    ) as rank_in_year
FROM eac_indicators
WHERE "Series Name" = 'Gross national expenditure (% of GDP)'
AND "Value" IS NOT NULL
ORDER BY "Year", rank_in_year
"""

result11 = pd.read_sql_query(query11, conn)

# Show only Somalia's ranking across years
somalia_rank = result11[result11["Country Name"] == "Somalia, Fed. Rep."]
somalia_rank
```

```python
query12 = """
SELECT 
    i."Year",
    i."Value" AS inflation_rate,
    g."Value" AS gdp_per_capita
FROM eac_indicators i
JOIN eac_indicators g
    ON i."Country Name" = g."Country Name"
    AND i."Year" = g."Year"
WHERE i."Series Name" = 'Inflation, consumer prices (annual %)'
AND g."Series Name" = 'GDP per capita (current LCU)'
AND i."Country Name" = 'South Sudan'
AND i."Value" > 50
ORDER BY i."Year"
"""

result12 = pd.read_sql_query(query12, conn)
result12
```

```python
query13 = """
SELECT 
    u."Country Name",
    u."Year",
    u."Value" AS unemployment,
    i."Value" AS inflation
FROM eac_indicators u
JOIN eac_indicators i
    ON u."Country Name" = i."Country Name"
    AND u."Year" = i."Year"
WHERE u."Series Name" = 'Unemployment, total (% of total labor force) (national estimate)'
AND i."Series Name" = 'Inflation, consumer prices (annual %)'
AND u."Value" IS NOT NULL
AND i."Value" IS NOT NULL
ORDER BY u."Country Name", u."Year"
"""

result13 = pd.read_sql_query(query13, conn)
result13
```

```python
# Calculate correlation between unemployment and inflation
# For all countries combined
overall_corr = result13["unemployment"].corr(result13["inflation"])
print(f"Overall correlation: {overall_corr:.3f}")

# Correlation by country
country_corr = result13.groupby("Country Name")[["unemployment", "inflation"]].corr()
print("\nCorrelation by country:")
print(result13.groupby("Country Name").apply(
    lambda x: x["unemployment"].corr(x["inflation"])
).round(3))
```

```python
query14 = """
SELECT 
    "Year",
    "Value" AS expenditure,
    LAG("Value") OVER (ORDER BY "Year") AS prev_year_expenditure,
    "Value" - LAG("Value") OVER (ORDER BY "Year") AS annual_change
FROM eac_indicators
WHERE "Series Name" = 'Gross national expenditure (% of GDP)'
AND "Country Name" = 'Somalia, Fed. Rep.'
AND "Value" IS NOT NULL
ORDER BY annual_change DESC
"""

result14 = pd.read_sql_query(query14, conn)
result14
```
