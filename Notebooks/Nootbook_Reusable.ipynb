```python
def clean_world_bank_data(filepath, missing_codes=[".."],
                           start_year=2010, end_year=2024,
                           save_path=None):
    
    # Block 1 - Open the file
    df = pd.read_excel(filepath, sheet_name="Data", dtype=str)
    
    # Block 2 - Remove metadata rows
    valid_countries = [
        "Somalia, Fed. Rep.", "Kenya", "Uganda",
        "South Sudan", "Rwanda", "Congo, Dem. Rep.",
        "Tanzania", "Burundi"
    ]
    df = df[df["Country Name"].isin(valid_countries)]
    df = df.dropna(subset=["Country Name"])
    
    # Block 3 - Replace missing codes and fix dtypes
    for code in missing_codes:
        df = df.replace(code, np.nan)
    
    year_cols = [col for col in df.columns if "YR" in col]
    df[year_cols] = df[year_cols].apply(pd.to_numeric, errors="coerce")
    
    # Block 4 - Clean column names and drop outside year range
    df.columns = [col.split(" ")[0] if "YR" in col else col for col in df.columns]
    
    all_years = [str(y) for y in range(2010, 2026)]
    keep_years = [str(y) for y in range(start_year, end_year + 1)]
    drop_years = [y for y in all_years if y not in keep_years]
    df = df.drop(columns=[y for y in drop_years if y in df.columns])
    
    # Block 5 - Reshape from wide to long
    keep_cols = ["Country Name", "Country Code", "Series Name", "Series Code"]
    value_cols = [str(y) for y in range(start_year, end_year + 1)]
    
    df_long = pd.melt(
        df,
        id_vars=keep_cols,
        value_vars=value_cols,
        var_name="Year",
        value_name="Value"
    )
    
    df_long["Year"] = df_long["Year"].astype(int)
    
    # Block 6 - Summary and save
    total_cells = df_long["Value"].shape[0]
    missing_cells = df_long["Value"].isna().sum()
    missing_pct = round((missing_cells / total_cells) * 100, 1)
    
    print("=" * 50)
    print("CLEANING SUMMARY")
    print("=" * 50)
    print(f"Final shape      : {df_long.shape}")
    print(f"Total cells      : {total_cells}")
    print(f"Missing values   : {missing_cells} ({missing_pct}%)")
    print("\nMissing by Country-Indicator:")
    missing_summary = (
        df_long[df_long["Value"].isna()]
        .groupby(["Country Name", "Series Name"])
        .size()
        .reset_index(name="Missing Years")
    )
    print(missing_summary.to_string(index=False))
    
    if save_path:
        df_long.to_csv(save_path, index=False)
        print(f"\nSaved to: {save_path}")
    print("=" * 50)
    
    return df_long
```
